```python
import torch
import numpy as np
import matplotlib.pyplot as plt

torch.manual_seed(42)
print('PyTorch version:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
```

## 1. Tensor Creation & Operations

```python
# Creation
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.randn(3, 4)
c = torch.zeros(2, 2)
d = torch.arange(0, 10, 2)
print('a:', a)
print('b shape:', b.shape)
print('d:', d)

# Reshape / transpose
e = b.view(4, 3)
print('e shape:', e.shape)
print('b.T shape:', b.T.shape)

# Element-wise ops
print('a + 10:', a + 10)
print('a * b[0]:', a * b[0])
print('dot:', torch.dot(a, a))

# Reduction
print('sum:', b.sum().item())
print('mean:', b.mean().item())
print('max:', b.max().item())
```

## 2. Autograd — Automatic Differentiation

```python
# y = x^2 + 3x + 2 ; dy/dx = 2x + 3
x = torch.tensor(4.0, requires_grad=True)
y = x**2 + 3*x + 2
y.backward()
print('x =', x.item())
print('y =', y.item())
print('dy/dx =', x.grad.item(), '(expected:', 2*4.0 + 3, ')')
```

```python
# Gradient descent on y = (x - 3)^2
x = torch.tensor(0.0, requires_grad=True)
lr = 0.1
losses = []
for i in range(50):
    loss = (x - 3)**2
    loss.backward()
    with torch.no_grad():
        x -= lr * x.grad
    x.grad.zero_()
    losses.append(loss.item())

print('Final x:', x.item())
plt.plot(losses)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Gradient Descent Convergence')
plt.savefig('outputs/autograd_convergence.png', dpi=300, bbox_inches='tight')
plt.show()
```

## 3. Loss Functions & Optimisers

```python
# MSE Loss
pred = torch.tensor([1.0, 2.0, 3.0])
target = torch.tensor([1.5, 2.5, 2.5])
mse = torch.nn.MSELoss()(pred, target)
print('MSE:', mse.item())

# Cross-entropy loss
logits = torch.randn(4, 3)
labels = torch.tensor([0, 1, 2, 1])
ce = torch.nn.CrossEntropyLoss()(logits, labels)
print('CrossEntropy:', ce.item())

# Optimisers
params = torch.nn.Parameter(torch.randn(3, 3))
sgd = torch.optim.SGD([params], lr=0.01)
adam = torch.optim.Adam([params], lr=0.001)
print('SGD:', sgd)
print('Adam:', adam)
```
