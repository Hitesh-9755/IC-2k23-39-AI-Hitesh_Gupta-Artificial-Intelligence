```python
# --- 2. Frames ---
class Frame:
    def __init__(self, name):
        self.name = name
        self.slots = {}
    def set(self, slot, value):
        self.slots[slot] = value
    def get(self, slot, default=None):
        return self.slots.get(slot, default)
    def __repr__(self):
        return f"Frame({self.name!r}, slots={self.slots})"

course = Frame('CS101')
course.set('title', 'Introduction to AI')
course.set('credits', 4)
course.set('department', 'CSE')
course.set('prerequisite', 'MA101')
course.set('instructor', 'DrSmith')
print(course)
print('Prerequisite:', course.get('prerequisite'))
```

```python
# --- 3. Rule Engine (Medical KB) ---
med = build_medical_engine()
trace = med.run()
print('Firing trace:')
for step in trace:
    print(' ', step)
print('\nDerived facts:')
for fact in sorted(med.all_facts(), key=str):
    print(' ', fact)
```

```python
# --- 4. Rule Engine (University KB) ---
uni = build_university_engine()
uni.run()
print('University KB facts:')
for fact in sorted(uni.all_facts(), key=str):
    print(' ', fact)
```

## Comparison: Frames vs. Semantic Networks vs. Rule-Based Systems

| Aspect | Frames | Semantic Networks | Rule-Based Systems |
|--------|--------|-------------------|--------------------|
| Structure | Slot-value pairs | Graph (nodes+edges) | Facts + IF-THEN rules |
| Inheritance | Via frame hierarchy | Via IS-A links | Via rule chaining |
| Reasoning | Property inheritance | Graph traversal | Forward/backward chaining |
| Best for | Object description | Concept hierarchies | Diagnostic/decision logic |
| Flexibility | Moderate | High | High |
| Complexity | Low | Medium | Medium-High |
