This notebook demonstrates BFS, DFS, Hill Climbing, Best-First Search,
and A* on a weighted route-planning graph, and compares their performance.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from Lab2.SearchAlgorithms import (
    build_route_graph, bfs, dfs, hill_climbing, best_first, astar,
    straight_line_heuristic
)
import matplotlib.pyplot as plt
import pandas as pd

graph = build_route_graph()
start, goal = 'Arad', 'Bucharest'
h_dist = {
    'Arad': 366, 'Zerind': 374, 'Oradea': 380, 'Sibiu': 253,
    'Fagaras': 176, 'RimnicuVilcea': 193, 'Timisoara': 329,
    'Lugoj': 244, 'Mehadia': 241, 'Drobeta': 242,
    'Craiova': 160, 'Pitesti': 100, 'Bucharest': 0,
}
heuristic = straight_line_heuristic(h_dist, goal)
```

## Run all algorithms and collect metrics

```python
algorithms = {
    'BFS': lambda: bfs(graph, start, goal),
    'DFS': lambda: dfs(graph, start, goal),
    'Hill Climbing': lambda: hill_climbing(graph, start, goal, heuristic),
    'Best-First': lambda: best_first(graph, start, goal, heuristic),
    'A*': lambda: astar(graph, start, goal, heuristic),
}

rows = []
for name, fn in algorithms.items():
    path, expanded, elapsed = fn()
    rows.append({
        'Algorithm': name,
        'Solved': path is not None,
        'Path Length': len(path) - 1 if path else None,
        'Nodes Expanded': expanded,
        'Time (ms)': round(elapsed * 1000, 3),
    })
    if path:
        print(f"{name:<15} {' -> '.join(path)}")

df = pd.DataFrame(rows)
df
```

```python
# Bar chart: nodes expanded per algorithm
ax = df.plot.bar(x='Algorithm', y='Nodes Expanded', legend=False,
                 figsize=(8, 4), color='steelblue')
ax.set_ylabel('Nodes Expanded')
ax.set_title('Search Algorithm Comparison — Nodes Expanded')
plt.tight_layout()
plt.savefig('outputs/search_comparison.png', dpi=300)
plt.show()
```
